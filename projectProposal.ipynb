# Project: Basketball Players Stats per Season - 49 Leagues

### Harrison Carter, Kyle Hays

## Dataset Description: 

[Project Link](https://www.kaggle.com/datasets/jacobbaruch/basketball-players-stats-per-season-49-leagues)

This CVS formated Database has 34 attributes regarding the  League, Season, Player, Team, Seasion Stats, Player Details (birth date, nationality, height, weight, draft). With the key being Season, Player, Team.

The outcome we are predicting is Player performance, where we will take a player's stats and categorize them into Elite, Above Average, Average, and below average. This will be based on the season statistics of Min played, Field goals made, 3-pointers made, rebounds, steals, true shooting percentage, and turnovers.

## Implementation/technical merit:

One potential problem would be regarding determining what would count to each category. There is a chance there are extremely large numbers in regards to stats, however we don't think they will be outliers in the data but if they are extremely high we may remove or reduce the number to the next highest although we doubt that will be needed.

## Potential Impact of the results:

These results would be useful for identifying players that are performing at high levels and determining whether players are performing well according to the numbers alone or if they are perfoming well despite numbers stating otherwise purely off of a "looks test" to the average viewer. 
Some stakeholders that might be intersted in these reuslts are leagues and teams that use player data to track if they are performing well and if they are what that specified team/league may be looking for in a player based on their statistics.
